# TensorFlow를 활용한 선형회귀 실습

이 노트북에서는 TensorFlow를 사용하여 간단한 선형회귀 문제를 풀어봅니다.

## 주요 단계
1. 데이터 생성 (y = 3x + 2 + 잡음)
2. 모델 정의 (Dense Layer)
3. 손실 함수 및 최적화 방법 설정
4. 학습 진행 및 시각화


```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print(tf.__version__)
```

## 1. 데이터 생성

```python
# y = 3x + 2 + noise
n_samples = 5
X = np.linspace(-5, 5, n_samples)
y = 3 * X + 2 + np.random.normal(0, 1, n_samples)*20

plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Generated Data")
plt.show()
```

## 2. 모델 정의

```python
model = tf.keras.Sequential([
    tf.keras.layers.InputLayer(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.summary()
```

## 3. 모델 컴파일
- 손실 함수: MSE (Mean Squared Error)
- 최적화 방법: SGD (Stochastic Gradient Descent)

```python
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              loss='mse')
```

## 4. 모델 학습

```python
history = model.fit(X, y, epochs=200, verbose=2)

plt.plot(history.history['loss'])
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('Training Loss')
plt.show()
```

## 5. 결과 확인

```python
W, b = model.layers[0].get_weights()
print(f"학습된 기울기 W: {W[0][0]:.3f}, 절편 b: {b[0]:.3f}")

y_pred = model.predict(X)

plt.scatter(X, y, label='Data')
plt.plot(X, y_pred, color='red', label='Fitted Line')
plt.legend()
plt.show()
```

## 생각해보기
- 학습 샘플 수가 작아지면 어떻게 되는가?

**답변**

샘플 수가 줄어들면 **파라미터 추정의 분산(variance)이 커져서, 학습할 때마다 전혀 다른 직선이 나오고 참값에서 크게 벗어난다.**

**이번 실행 결과가 그대로 보여줌** — `n_samples = 5`로 학습한 결과:

| 파라미터 | 참값 (데이터 생성식) | 학습된 값 | 오차 |
|---|---|---|---|
| 기울기 W | 3 | **3.531** | +17.7% |
| 절편 b | 2 | **7.224** | **+261%** |

기울기는 그런대로 맞췄지만 절편은 3.6배로 틀렸다. 절편 추정은 y값의 평균에 직접 의존하는데, 표준편차 20짜리 잡음이 섞인 값 5개의 평균은 참 평균에서 쉽게 ±9 정도 벗어나기 때문이다.

**왜 그런가**

- 파라미터는 2개(W, b)인데 데이터는 5점뿐이라 **자유도가 3밖에 없다.** 잡음을 평균으로 상쇄시킬 여유가 없어서, 잡음이 그대로 파라미터에 흡수된다.
- 최소제곱 추정의 표준오차는 `SE(W) ≈ σ / √(Σ(xᵢ - x̄)²)` 이고, x가 같은 구간에 퍼져 있다면 분모는 n에 비례한다. 즉 **추정 오차는 대략 1/√n 으로만 줄어든다.**
- 이 노트북은 `np.random.normal`에 seed가 없으므로 **다시 실행할 때마다 W, b가 크게 요동친다.** 이 불안정성 자체가 곧 "분산이 크다"는 뜻이다.

**극단적인 경우**

- n = 2 : 두 점을 정확히 지나는 직선이 유일하게 존재 → **학습 손실은 0이 되지만 잡음까지 완벽히 외운 것**이라 새 데이터에는 전혀 맞지 않는다. 전형적인 과적합.
- n = 1 : 직선이 무수히 많아 해가 결정되지 않는다.

**강의 내용과의 연결** — 3주차 강의 33p의 *"과적합 문제가 발생하는 이유1: 학습 데이터의 다양성 부족"* 이 바로 이 상황이다. 데이터가 x ∈ [-5, 5] 구간을 5점으로만 듬성듬성 덮고 있어서, 모델이 학습 구간의 빈 곳에서 어떻게 행동할지 보장할 수 없다.

- 학습 샘플에 잡음이 많으면 어떻게 되는가? 이 경우 학습 데이터가 많이 필요한가?

**답변**

**① 잡음이 많으면 → 손실이 0까지 내려가지 않고, 파라미터 추정이 부정확해진다.**

이번 실행의 학습 곡선을 보면 MSE가 **287.6 근처에서 평평해지고 더 이상 내려가지 않는다.**

| Epoch | 1 | 50 | 100 | 150 | 200 |
|---|---|---|---|---|---|
| Loss (MSE) | 471.8 | 295.1 | 288.6 | 287.8 | **287.6** |

이건 학습이 실패한 게 아니다. 데이터를 `y = 3x + 2 + N(0, 1)×20` 으로 만들었으니 **잡음 분산이 20² = 400**이고, 아무리 완벽한 직선을 찾아도 MSE를 잡음 분산 아래로 내릴 수 없다. 이 부분이 **줄일 수 없는 오차(irreducible error)** 다.

참고로 신호와 잡음의 크기를 비교하면 — x가 [-5, 5]일 때 신호 `3x`의 변동 폭은 30인데 잡음의 표준편차가 20이다. **신호 대 잡음비가 매우 나쁜 데이터**이고, 그래서 산점도만 봐서는 직선 관계가 잘 보이지도 않는다.

**② 그래서 학습 데이터가 많이 필요한가? → 네, 그것도 훨씬 많이 필요하다.**

앞 문항의 `SE(W) ≈ σ / √(Σ(xᵢ - x̄)²)` 를 다시 보면, 추정 오차는 **잡음 σ에 비례하고 √n에 반비례**한다. 정리하면:

> 같은 추정 정확도를 유지하려면 **필요한 샘플 수 n은 잡음 분산 σ²에 비례한다.**
> 즉 **잡음이 2배가 되면 데이터는 4배가 필요하다.**

이번 실습은 잡음이 σ=20으로 매우 큰데 샘플은 5개뿐이라, 두 악조건이 겹쳐 절편이 3.6배나 틀린 결과가 나왔다.

**③ 다만 "데이터를 늘리는 것"이 유일한 해법은 아니다**

- 데이터가 잡음에 비해 부족한 상태에서 모델을 키우면, 모델이 **잡음을 신호로 착각해서 학습**한다(과적합). 잡음이 큰 데이터일수록 단순한 모델이 유리하다.
- 규제화(L1/L2)로 파라미터가 잡음을 따라가지 못하도록 제약을 거는 방법도 있다. *(4주차 Learning Strategies에서 다룸)*
- 애초에 σ 자체를 줄이는 것이 가장 효과적이다. 제조 현장이라면 **센서 교정, 반복 측정 후 평균, 필터링**이 여기에 해당한다. n을 4배 늘리는 것보다 σ를 절반으로 줄이는 쪽이 대개 더 싸다.
